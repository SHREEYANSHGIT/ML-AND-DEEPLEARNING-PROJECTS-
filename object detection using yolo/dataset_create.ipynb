```python
import cv2
import os

# ---------------- CONFIG ----------------
SAVE_DIR = "dataset"
CLASSES = {
    "p": "Person",
    "b": "pen",
    "m": "facewash"
}
IMAGE_SIZE = 720
# --------------------------------------

# Create directories if not exist
for cls in CLASSES.values():
    os.makedirs(os.path.join(SAVE_DIR, cls), exist_ok=True)

# Count existing images (resume-safe)
img_count = {
    cls: len(os.listdir(os.path.join(SAVE_DIR, cls)))
    for cls in CLASSES.values()
}

# cap = cv2.VideoCapture("http://192.168.29.227:4747/video")

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("❌ Webcam not accessible")
    exit()

print("🎥 Webcam started")
print("Controls:")
print(" p → save PERSON")
print(" b → save pen")
print(" m → save facewash")
print(" q → quit")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.resize(frame, (IMAGE_SIZE, IMAGE_SIZE))

    # ----------- TEXT OVERLAY -----------
    cv2.putText(frame, "p:Person  b:pen  m:facewash  q:Quit",
                (10, 30), cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (0, 255, 0), 2)

    cv2.putText(frame, f"Person : {img_count['Person']}",
                (10, 70), cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (255, 255, 0), 2)

    cv2.putText(frame, f"pen : {img_count['pen']}",
                (10, 100), cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (255, 255, 0), 2)

    cv2.putText(frame, f"facewash  : {img_count['facewash']}",
                (10, 130), cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (255, 255, 0), 2)
    # ------------------------------------

    cv2.imshow("Dataset Capture", frame)

    key = cv2.waitKey(1) & 0xFF

    if key == ord('q'):
        break

    elif chr(key) in CLASSES:
        cls_name = CLASSES[chr(key)]
        img_name = f"{cls_name}_{img_count[cls_name]:05d}.jpg"
        save_path = os.path.join(SAVE_DIR, cls_name, img_name)

        cv2.imwrite(save_path, frame)
        img_count[cls_name] += 1

        print(f"✅ Saved {cls_name}: {img_count[cls_name]}")

cap.release()
cv2.destroyAllWindows()
print("🛑 Capture stopped")
```
