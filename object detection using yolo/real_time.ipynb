```python
import cv2
from ultralytics import YOLO
import time

# Load trained model
model = YOLO("runs/detect/train/weights/best.pt").to("cuda")

# Open IP webcam (DroidCam / IP Webcam)
# cap = cv2.VideoCapture("http://192.168.29.227:4747/video")
# For laptop webcam use:
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("❌ Camera not opening")
    exit()

prev_time = 0

while True:
    ret, frame = cap.read()
    if not ret:
        print("❌ Frame not received")
        break

    # YOLO inference
    results = model(frame, conf=0.3, stream=True)

    for r in results:
        annotated_frame = r.plot()

    # FPS calculation
    curr_time = time.time()
    fps = 1 / (curr_time - prev_time) if prev_time != 0 else 0
    prev_time = curr_time

    # Draw FPS
    cv2.putText(
        annotated_frame,
        f"FPS: {int(fps)}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    cv2.imshow("YOLOv8 Real-Time Detection", annotated_frame)

    # Press 'q' to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
